## Бизнес задача

**Проблема:** Футбольные клубы часто переплачивают за игроков или упускают таланты, потому что оценка стоимости субъективна и зависит от агентов и хайпа.

**Цель:** Предсказать справедливую рыночную стоимость футболиста на основе объективных данных (статистика, возраст, позиция), чтобы скауты могли быстро находить недооценённых игроков и избегать переплат.



## ML задача

**Тип задачи:** Регрессия (предсказываем число).

**Таргет:** `current_value_eur` — рыночная стоимость игрока в евро.

**Признаки:**
- **Демография:** возраст, позиция, гражданство.
- **Карьера:** текущий клуб, лига, опыт (лет в проф. футболе), позиция на поле
- **Динамика:** как быстро росла стоимость (`value_cagr`), была ли она выше в прошлом, действующие контракты
- **Статистика:** матчи, голы, ассисты, карточки в играх




## Набор данных

**Источник:** [Football Transfer Value Intelligence 2024 (Kaggle)](https://www.kaggle.com/datasets/kanchana1990/football-transfer-value-intelligence-2024)

**Что внутри:**
- 508 игроков из топ-5 европейских лиг (АПЛ, Ла Лига, Бундеслига, Серия А, Лига 1).
- 38 признаков: от возраста и позиции до динамики стоимости и карьерной траектории.
- Целевая переменная: `current_value_eur` (от €50 тыс. до €180 млн).

**Качество данных:**
- Пропусков мало (<4%)
- Данные актуальны (2020–2024 гг.).


## Метрика

**Основная метрика:** MAPE (Mean Absolute Percentage Error) — средняя ошибка в процентах.

**Дополнительная:** Spearman correlation — насколько правильно модель ранжирует игроков (кто дороже, кто дешевле).

**Обоснование:**
1. **Понятна бизнесу.** Легче объяснить скауту «ошибка ~20%», чем «MSE = 10^12».
2. **Устойчива к выбросам.** В футболе есть игроки за €100+ млн, которые перетягивают абсолютные метрики. Процентная ошибка уравнивает вклад дешёвых и дорогих игроков.
3. **Важность ранжирования.** Скаутам часто важнее понять, кто перспективнее относительно других, чем знать точную цифру до евро. Поэтому ещё и Spearman.

# Часть 2: Бейзлайн-модель

## Разбиение на train/test
Для оценки обобщающей способности модели используем стратифицированное (по позициям) или случайное разбиение 80/20.
Фиксируем `random_state=42` для полной воспроизводимости результатов от запуска к запуску.

```python
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.metrics import mean_absolute_percentage_error
from scipy.stats import spearmanr
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
import shap
import warnings


warnings.filterwarnings('ignore')
RANDOM_STATE = 42


X = df.drop(columns=['current_value_eur'])
y = df['current_value_eur']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

print(f"Train: {X_train.shape[0]} записей, Test: {X_test.shape[0]} записей")
```

## Константное предсказание
В качестве наивного бейзлайна предсказываем медиану целевой переменной по тренировочной выборке.
Это позволяет оценить, насколько лучше работает модель по сравнению с простым угадыванием «типичной» стоимости.

```python
# медиана train-выборки
median_pred = np.full_like(y_test, fill_value=y_train.median(), dtype=float)

mape_baseline = mean_absolute_percentage_error(y_test, median_pred)
spearman_baseline, _ = spearmanr(y_test, median_pred)

print(f"📉 Константный бейзлайн (медиана):")
print(f"   MAPE: {mape_baseline:.2%}")
print(f"   Spearman: {spearman_baseline:.3f}")
```

## Простая модель: Ridge-регрессия
**Обоснование выбора и предобработки:**
1. **Утечка данных:** В EDA выявлена сильная корреляция `current_value_eur` ↔ `peak_value_eur` (0.9) и признаки `value_cagr`, `ever_*`, которые рассчитываются с использованием целевой переменной. Чтобы модель училась прогнозировать, а не запоминать, исключаем все производные от таргета признаки.
2. **Логарифмирование таргета:** Распределение стоимости сильно скошено (skew > 2). Логарифмирование `log1p(y)` стабилизирует дисперсию и делает задачу ближе к линейной.
3. **Кодирование категорий:** `position_group`, `league_name`, `nationality`, `trajectory` кодируем через One-Hot Encoding (`drop_first=True`), так как количество уникальных значений невелико, а линейная модель не умеет работать с категориями напрямую.
4. **Модель:** `Ridge` (L2-регуляризация) предотвращает переобучение на разреженных OHE-признаках и устойчив к мультиколлинеарности.

```python
# 1. Отбор признаков (исключаем утечки и избыточные)
drop_cols = [
    'player_id', 'name', 'current_value_tier', 'peak_value_eur', 'peak_value_tier',
    'peak_date', 'peak_club', 'age_at_peak', 'first_value_eur', 'first_date',
    'last_value_eur', 'last_date', 'value_cagr', 'value_to_peak_cagr',
    'value_multiplier_x', 'post_peak_decline_pct', 'value_volatility',
    'mean_yoy_growth_rate', 'ever_100m', 'ever_50m', 'ever_10m',
    'data_source', 'dataset_built_at', 'current_value_eur'
]
feature_cols = [c for c in X_train.columns if c not in drop_cols]

# 2. Заполнение пропусков ПЕРЕД кодированием
# Числовые признаки - медианой
num_features = ['age', 'career_span_years', 'years_to_peak', 'num_valuation_points', 'num_clubs_career']
for col in num_features:
    if col in X_train.columns:
        median_val = X_train[col].median()
        X_train[col] = X_train[col].fillna(median_val)
        X_test[col] = X_test[col].fillna(median_val)

# Категориальные признаки - модой
cat_features = ['nationality', 'position_group', 'league_name', 'current_club', 'trajectory']
for col in cat_features:
    if col in X_train.columns:
        mode_val = X_train[col].mode()[0]
        X_train[col] = X_train[col].fillna(mode_val)
        X_test[col] = X_test[col].fillna(mode_val)

# 3. Логарифмирование таргета
y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)

# 4. One-Hot Encoding
X_train_enc = pd.get_dummies(X_train[feature_cols], drop_first=True)
X_test_enc = pd.get_dummies(X_test[feature_cols], drop_first=True)

# Выравнивание колонок после get_dummies
X_train_enc, X_test_enc = X_train_enc.align(X_test_enc, join='left', axis=1, fill_value=0)

print(f"Пропуски заполнены. Признаков после кодирования: {X_train_enc.shape[1]}")

# 5. Обучение Ridge
ridge_model = Ridge(alpha=1.0, random_state=RANDOM_STATE)
ridge_model.fit(X_train_enc, y_train_log)

# 6. Прогноз и метрики
y_pred_log = ridge_model.predict(X_test_enc)
y_pred = np.expm1(y_pred_log)

mape_ridge = mean_absolute_percentage_error(y_test, y_pred)
spearman_ridge, _ = spearmanr(y_test, y_pred)

print(f"\nRidge (log-target):")
print(f"   MAPE: {mape_ridge:.2%}")
print(f"   Spearman: {spearman_ridge:.3f}")
```

# Ансамблевая модель и оптимизация гиперпараметров

## Выбор модели и стратегии tuning
- **Модель:** `RandomForestRegressor` — устойчив к выбросам, не требует масштабирования, автоматически обрабатывает нелинейные зависимости (возраст→стоимость, лига→премиум).
- **Поиск гиперпараметров:** `RandomizedSearchCV` с 5-fold кросс-валидацией. Оптимизируем `n_estimators`, `max_depth`, `min_samples_split`, `min_samples_leaf`. Эти параметры контролируют сложность деревьев и риск переобучения на выборке из 508 игроков.
- **Метрика оптимизации:** `neg_mean_absolute_percentage_error` (максимизация отрицательного MAPE эквивалентна минимизации MAPE).

```python
# Подготовка данных для ансамбля (те же признаки, сырой таргет)
rf_X_train = X_train_enc.copy()
rf_X_test = X_test_enc.copy()

# Сетка гиперпараметров
param_dist = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 5, 7, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf = RandomForestRegressor(random_state=RANDOM_STATE)

# RandomizedSearchCV
search = RandomizedSearchCV(
    estimator=rf,
    param_distributions=param_dist,
    n_iter=15,
    cv=5,
    scoring='neg_mean_absolute_percentage_error',
    random_state=RANDOM_STATE,
    n_jobs=-1,
    verbose=1
)

search.fit(rf_X_train, y_train)
best_rf = search.best_estimator_

print(f"\nЛучшие гиперпараметры: {search.best_params_}")

# Оценка на тесте
y_pred_rf = best_rf.predict(rf_X_test)
mape_rf = mean_absolute_percentage_error(y_test, y_pred_rf)
spearman_rf, _ = spearmanr(y_test, y_pred_rf)

print(f"\nRandom Forest (тест):")
print(f"   MAPE: {mape_rf:.2%}")
print(f"   Spearman: {spearman_rf:.3f}")
```

# Интерпретация модели

## Глобальная интерпретация
Используем **Permutation Importance** (scikit-learn), так как он показывает реальное влияние признака на метрику MAPE, не завися от масштаба данных и корреляций между фичами. Дополнительно выводим встроенные `feature_importances_` дерева для сравнения.

```python
# Permutation Importance
perm_imp = permutation_importance(
    best_rf, rf_X_test, y_test,
    n_repeats=10,
    random_state=RANDOM_STATE,
    scoring='neg_mean_absolute_percentage_error',
    n_jobs=-1
)



# Встроенная важность
importances = pd.Series(best_rf.feature_importances_, index=rf_X_train.columns).sort_values(ascending=False)
print("\nТоп-10 признаков (Gini importance):")
print(importances.head(10))
```

## Локальная интерпретация
Применяем **SHAP (TreeExplainer)** для разбора предсказаний на конкретных игроках из тестовой выборки. Показываем, как каждый признак сдвигает прогноз относительно базового значения (среднего по train).

```python
# SHAP объяснение для 3 случайных игроков
explainer = shap.TreeExplainer(best_rf)
shap_values = explainer.shap_values(rf_X_test)

# Визуализация waterfall для первого игрока теста
shap.initjs()
shap.waterfall_plot(
    shap.Explanation(
        values=shap_values[0],
        base_values=explainer.expected_value,
        data=rf_X_test.iloc[0].values,
        feature_names=rf_X_test.columns
    ),
    max_display=10
)
```

## Экспертное мнение
Модель адекватно отражает экономику трансферного рынка: позиция и лига дают основной вклад в стоимость, возраст работает нелинейно (пик 24–27 лет), а исключение производных признаков предотвратило запоминание «ценника» вместо прогнозирования. Локальные разборы SHAP показывают, что переоценка/недооценка конкретного игрока объясняется именно отклонением его профиля от среднего по когорте (например, молодой игрок из АПЛ с траекторией `rising_star` получает надбавку к базовой стоимости).
